# 手势控制机械臂（P4 抓取）

PC 端运行 `手势识别` 项目，机械臂主机运行本 notebook。

## 使用前

1. 确认机械臂上电，`config.py` 的串口正确。
2. 清空机械臂运动范围，确认 P4 物块已放好。
3. 机械臂主机与 PC 在同一网段，PC 可访问本机 `8765` 端口。

## 手势

- `ONE`：识别 P4 色块并进入待确认状态，**不运动**。
- `TWO`：选择“复位到拍摄位”，**不运动**。
- `OK`：确认并执行一次已选择动作。
- `OPEN_PALM`：取消待确认动作。
- `FIST`：急停并卸力。

> 首次运行会先把机械臂移动到 `capture_image` 拍摄位。请先确保周围无人、无障碍物。
